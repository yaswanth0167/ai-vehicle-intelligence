```python
repair_df["Towing Date"] = repair_df["Towing Date"].fillna("No Towing")
```

```python
repair_df["Tow Location"] = repair_df["Tow Location"].fillna("Not Applicable")
```

```python
repair_df["Drop-off Location"] = repair_df["Drop-off Location"].fillna("Not Applicable")
```

```python
repair_df["Tow Truck Driver"] = repair_df["Tow Truck Driver"].fillna("Not Assigned")
```

```python
repair_df.drop("Customer Feedback", axis=1, inplace=True)
```

```python
current_year = 2026
used_car_df["Age"] = current_year - used_car_df["Year"]
```

```python
used_car_df["Age"] = used_car_df["Age"].fillna(used_car_df["Age"].median())
```

```python
used_car_df["Transmission"] = used_car_df["Transmission"].fillna(
    used_car_df["Transmission"].mode()[0]
)
```

```python
used_car_df["FuelType"] = used_car_df["FuelType"].fillna(
    used_car_df["FuelType"].mode()[0]
)
```

```python
# 1. Strip 'km' and commas, then convert to numeric
used_car_df["kmDriven"] = (
    used_car_df["kmDriven"]
    .astype(str)
    .str.replace("km", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

used_car_df["kmDriven"] = pd.to_numeric(used_car_df["kmDriven"], errors="coerce")

# 2. Fill missing values with the median
used_car_df["kmDriven"] = used_car_df["kmDriven"].fillna(
    used_car_df["kmDriven"].median()
)
```

```python
diagnosis_df["ECU Data"] = diagnosis_df["ECU Data"].fillna("Unknown")
```

```python
diagnosis_df["Service History"] = diagnosis_df["Service History"].fillna("No History")
```
