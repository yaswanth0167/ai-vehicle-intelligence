```python
print("========== LOADING FRAUD DATASET ==========")

import pandas as pd
import numpy as np

fraud_df = pd.read_csv("Y:/AI_Vehicle_Service_Repair_Insurance_Intelligence/cleaned/fraud_clean.csv")

print("Dataset loaded successfully!")

print("\nDataset shape:")
print(fraud_df.shape)

print("\nFirst 5 rows:")
display(fraud_df.head())

print("\nColumn names:")
print(fraud_df.columns.tolist())
```

```python
print("========== FRAUD DATASET BASIC VALIDATION ==========")

print("Shape:")
print(fraud_df.shape)

print("\nData types:")
display(fraud_df.dtypes)

print("\nMissing values:")
display(
    fraud_df.isnull().sum()
    .sort_values(ascending=False)
)

print("\nDuplicate rows:")
print(fraud_df.duplicated().sum())

print("\nTarget distribution:")
display(
    fraud_df["FraudFound_P"].value_counts()
)

print("\nTarget percentages:")
display(
    fraud_df["FraudFound_P"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("========== FRAUD DATASET TARGET & FEATURE ANALYSIS ==========")

print("\nTarget distribution:")
display(
    fraud_df["FraudFound_P"].value_counts()
)

print("\nTarget percentages:")
display(
    fraud_df["FraudFound_P"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nNumerical features:")

numerical_features_fraud = fraud_df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print(numerical_features_fraud)

print("\nCategorical features:")

categorical_features_fraud = fraud_df.select_dtypes(
    include=["object"]
).columns.tolist()

print(categorical_features_fraud)

print("\nNumber of numerical features:",
      len(numerical_features_fraud))

print("Number of categorical features:",
      len(categorical_features_fraud))
```

```python
import pandas as pd

print("========== CREATING FRAUD FEATURE-ENGINEERED DATASET ==========")

# 0. Copy from base DataFrame
fraud_fe = fraud_df.copy()

# -----------------------------
# 1. Vehicle age conversion
# -----------------------------
vehicle_age_map = {
    "new": 0,
    "2 years": 2,
    "3 years": 3,
    "4 years": 4,
    "5 years": 5,
    "6 years": 6,
    "7 years": 7,
    "more than 7": 8,
}

fraud_fe["Vehicle_Age_Num"] = fraud_fe["AgeOfVehicle"].map(vehicle_age_map)

# -----------------------------
# 2. Policy-holder age midpoint
# -----------------------------
policy_age_map = {
    "16 to 17": 16.5,
    "18 to 20": 19,
    "21 to 25": 23,
    "26 to 30": 28,
    "31 to 35": 33,
    "36 to 40": 38,
    "41 to 50": 45.5,
    "51 to 65": 58,
    "over 65": 70,
}

fraud_fe["PolicyHolder_Age_Num"] = fraud_fe["AgeOfPolicyHolder"].map(
    policy_age_map
)

# -----------------------------
# 3. Claim timing difference
# -----------------------------
fraud_fe["Claim_Week_Difference"] = (
    fraud_fe["WeekOfMonthClaimed"] - fraud_fe["WeekOfMonth"]
)

# -----------------------------
# 4. Driver age vs policy age
# -----------------------------
fraud_fe["Age_PolicyHolder_Difference"] = (
    fraud_fe["Age"] - fraud_fe["PolicyHolder_Age_Num"]
)

# -----------------------------
# 5. Vehicle age vs driver age
# -----------------------------
fraud_fe["Age_Vehicle_Interaction"] = (
    fraud_fe["Age"] * fraud_fe["Vehicle_Age_Num"]
)

# -----------------------------
# 6. Driver rating interaction
# -----------------------------
fraud_fe["Age_DriverRating_Interaction"] = (
    fraud_fe["Age"] * fraud_fe["DriverRating"]
)

# -----------------------------
# 7. Policy duration indicator
# -----------------------------
fraud_fe["Short_Policy_Accident"] = (
    fraud_fe["Days_Policy_Accident"]
    .astype(str)
    .str.contains("8 to 15|15 to 30", regex=True)
    .astype(int)
)

# -----------------------------
# Summary & Verification
# -----------------------------
print("Feature engineering completed successfully!")

print("\nOriginal features:", fraud_df.shape[1])
print("New features added:", fraud_fe.shape[1] - fraud_df.shape[1])
print("Total features:", fraud_fe.shape[1])

print("\nNew features:")
for feature in [
    "Vehicle_Age_Num",
    "PolicyHolder_Age_Num",
    "Claim_Week_Difference",
    "Age_PolicyHolder_Difference",
    "Age_Vehicle_Interaction",
    "Age_DriverRating_Interaction",
    "Short_Policy_Accident",
]:
    print("-", feature)
```

```python
print("========== FRAUD ENGINEERED FEATURE VALIDATION ==========")

new_fraud_features = [
    "Vehicle_Age_Num",
    "PolicyHolder_Age_Num",
    "Claim_Week_Difference",
    "Age_PolicyHolder_Difference",
    "Age_Vehicle_Interaction",
    "Age_DriverRating_Interaction",
    "Short_Policy_Accident"
]

display(
    fraud_fe[new_fraud_features].describe()
)

print("\nMissing values:")
display(
    fraud_fe[new_fraud_features].isnull().sum()
)

print("\nInfinite values:")
print(
    np.isinf(
        fraud_fe[new_fraud_features]
        .select_dtypes(include=np.number)
    ).sum().sum()
)
```

```python
print("========== FRAUD FEATURE-ENGINEERED TRAIN TEST SPLIT ==========")

from sklearn.model_selection import train_test_split

# Separate features and target
X_fraud_fe = fraud_fe.drop(
    columns=["FraudFound_P"]
)

y_fraud_fe = fraud_fe["FraudFound_P"]

# Stratified train-test split
X_train_fraud_fe, X_test_fraud_fe, \
y_train_fraud_fe, y_test_fraud_fe = train_test_split(
    X_fraud_fe,
    y_fraud_fe,
    test_size=0.20,
    random_state=42,
    stratify=y_fraud_fe
)

print("Training data:")
print("X_train:", X_train_fraud_fe.shape)
print("y_train:", y_train_fraud_fe.shape)

print("\nTesting data:")
print("X_test :", X_test_fraud_fe.shape)
print("y_test :", y_test_fraud_fe.shape)

print("\nTraining target distribution:")
print(y_train_fraud_fe.value_counts())

print("\nTraining target percentages:")
print(
    y_train_fraud_fe.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nTesting target distribution:")
print(y_test_fraud_fe.value_counts())

print("\nTesting target percentages:")
print(
    y_test_fraud_fe.value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("========== CREATING FRAUD PREPROCESSING PIPELINE ==========")

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Remove target from feature lists
fraud_numerical_features = [
    "WeekOfMonth",
    "WeekOfMonthClaimed",
    "Age",
    "RepNumber",
    "Deductible",
    "DriverRating",
    "Year",
    "Vehicle_Age_Num",
    "PolicyHolder_Age_Num",
    "Claim_Week_Difference",
    "Age_PolicyHolder_Difference",
    "Age_Vehicle_Interaction",
    "Age_DriverRating_Interaction",
    "Short_Policy_Accident"
]

fraud_categorical_features = [
    "Month",
    "DayOfWeek",
    "Make",
    "AccidentArea",
    "DayOfWeekClaimed",
    "MonthClaimed",
    "Sex",
    "MaritalStatus",
    "Fault",
    "PolicyType",
    "VehicleCategory",
    "VehiclePrice",
    "Days_Policy_Accident",
    "Days_Policy_Claim",
    "PastNumberOfClaims",
    "AgeOfVehicle",
    "AgeOfPolicyHolder",
    "PoliceReportFiled",
    "WitnessPresent",
    "AgentType",
    "NumberOfSuppliments",
    "AddressChange_Claim",
    "NumberOfCars",
    "BasePolicy"
]

print("Numerical features:", len(fraud_numerical_features))
print("Categorical features:", len(fraud_categorical_features))

# Numerical pipeline
fraud_numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])

# Categorical pipeline
fraud_categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

# Combined preprocessing
fraud_preprocessor = ColumnTransformer([
    (
        "num",
        fraud_numeric_pipeline,
        fraud_numerical_features
    ),
    (
        "cat",
        fraud_categorical_pipeline,
        fraud_categorical_features
    )
])

print("Fraud preprocessing pipeline created successfully!")
```

```python
print("========== FITTING FRAUD PREPROCESSOR ==========")

X_train_fraud_processed = fraud_preprocessor.fit_transform(
    X_train_fraud_fe
)

print(
    "Processed training shape:",
    X_train_fraud_processed.shape
)
```

```python
print("========== TRANSFORMING FRAUD TEST DATA ==========")

X_test_fraud_processed = fraud_preprocessor.transform(
    X_test_fraud_fe
)

print(
    "Processed testing shape:",
    X_test_fraud_processed.shape
)
```

```python
print("========== FRAUD PREPROCESSING VALIDATION ==========")

print(
    "Original training features:",
    X_train_fraud_fe.shape[1]
)

print(
    "Processed training features:",
    X_train_fraud_processed.shape[1]
)

print(
    "\nOriginal testing features:",
    X_test_fraud_fe.shape[1]
)

print(
    "Processed testing features:",
    X_test_fraud_processed.shape[1]
)

print(
    "\nTraining samples:",
    X_train_fraud_processed.shape[0]
)

print(
    "Testing samples:",
    X_test_fraud_processed.shape[0]
)
```

```python
print("========== CREATING BASELINE FRAUD XGBOOST ==========")

from xgboost import XGBClassifier

xgb_fraud_baseline = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

print("Baseline Fraud XGBoost created successfully!")
```

```python
print("========== TRAINING BASELINE FRAUD XGBOOST ==========")

import time

start_time = time.time()

xgb_fraud_baseline.fit(
    X_train_fraud_processed,
    y_train_fraud_fe
)

training_time_fraud_baseline = time.time() - start_time

print("Training completed successfully!")
print(
    "Training time:",
    round(training_time_fraud_baseline, 2),
    "seconds"
)
```

```python
print("========== BASELINE FRAUD XGBOOST PREDICTIONS ==========")

y_pred_fraud_baseline = xgb_fraud_baseline.predict(
    X_test_fraud_processed
)

y_prob_fraud_baseline = xgb_fraud_baseline.predict_proba(
    X_test_fraud_processed
)[:, 1]

print("Predictions generated successfully!")

print("\nFirst 20 predictions:")
print(y_pred_fraud_baseline[:20])

print("\nFirst 20 fraud probabilities:")
print(y_prob_fraud_baseline[:20])
```

```python
print("========== BASELINE FRAUD XGBOOST EVALUATION ==========")

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

# Confusion matrix
cm_fraud_baseline = confusion_matrix(
    y_test_fraud_fe,
    y_pred_fraud_baseline
)

print("Confusion Matrix:")
print(cm_fraud_baseline)

# Classification metrics
precision_fraud_baseline = precision_score(
    y_test_fraud_fe,
    y_pred_fraud_baseline,
    zero_division=0
)

recall_fraud_baseline = recall_score(
    y_test_fraud_fe,
    y_pred_fraud_baseline,
    zero_division=0
)

f1_fraud_baseline = f1_score(
    y_test_fraud_fe,
    y_pred_fraud_baseline,
    zero_division=0
)

roc_auc_fraud_baseline = roc_auc_score(
    y_test_fraud_fe,
    y_prob_fraud_baseline
)

pr_auc_fraud_baseline = average_precision_score(
    y_test_fraud_fe,
    y_prob_fraud_baseline
)

print("\nPrecision:", round(precision_fraud_baseline, 4))
print("Recall   :", round(recall_fraud_baseline, 4))
print("F1 Score :", round(f1_fraud_baseline, 4))
print("ROC-AUC  :", round(roc_auc_fraud_baseline, 4))
print("PR-AUC   :", round(pr_auc_fraud_baseline, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test_fraud_fe,
        y_pred_fraud_baseline,
        target_names=["Non-Fraud", "Fraud"],
        zero_division=0
    )
)
```

```python
print("========== CREATING CLASS-WEIGHTED FRAUD XGBOOST ==========")

# Calculate fraud class weight
negative_count = (y_train_fraud_fe == 0).sum()
positive_count = (y_train_fraud_fe == 1).sum()

fraud_scale_pos_weight = negative_count / positive_count

print("Non-Fraud training cases:", negative_count)
print("Fraud training cases    :", positive_count)
print(
    "scale_pos_weight        :",
    round(fraud_scale_pos_weight, 4)
)

xgb_fraud_weighted = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=fraud_scale_pos_weight,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

print("Class-weighted Fraud XGBoost created successfully!")
```

```python
print("========== TRAINING CLASS-WEIGHTED FRAUD XGBOOST ==========")

start_time = time.time()

xgb_fraud_weighted.fit(
    X_train_fraud_processed,
    y_train_fraud_fe
)

training_time_fraud_weighted = time.time() - start_time

print("Training completed successfully!")
print(
    "Training time:",
    round(training_time_fraud_weighted, 2),
    "seconds"
)
```

```python
print("========== CLASS-WEIGHTED FRAUD XGBOOST PREDICTIONS ==========")

y_pred_fraud_weighted = xgb_fraud_weighted.predict(
    X_test_fraud_processed
)

y_prob_fraud_weighted = xgb_fraud_weighted.predict_proba(
    X_test_fraud_processed
)[:, 1]

print("Predictions generated successfully!")

print("\nFirst 20 predictions:")
print(y_pred_fraud_weighted[:20])

print("\nFirst 20 fraud probabilities:")
print(y_prob_fraud_weighted[:20])
```

```python
print("========== CLASS-WEIGHTED FRAUD XGBOOST EVALUATION ==========")

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

cm_fraud_weighted = confusion_matrix(
    y_test_fraud_fe,
    y_pred_fraud_weighted
)

print("Confusion Matrix:")
print(cm_fraud_weighted)

precision_fraud_weighted = precision_score(
    y_test_fraud_fe,
    y_pred_fraud_weighted,
    zero_division=0
)

recall_fraud_weighted = recall_score(
    y_test_fraud_fe,
    y_pred_fraud_weighted,
    zero_division=0
)

f1_fraud_weighted = f1_score(
    y_test_fraud_fe,
    y_pred_fraud_weighted,
    zero_division=0
)

roc_auc_fraud_weighted = roc_auc_score(
    y_test_fraud_fe,
    y_prob_fraud_weighted
)

pr_auc_fraud_weighted = average_precision_score(
    y_test_fraud_fe,
    y_prob_fraud_weighted
)

print("\nPrecision:", round(precision_fraud_weighted, 4))
print("Recall   :", round(recall_fraud_weighted, 4))
print("F1 Score :", round(f1_fraud_weighted, 4))
print("ROC-AUC  :", round(roc_auc_fraud_weighted, 4))
print("PR-AUC   :", round(pr_auc_fraud_weighted, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_test_fraud_fe,
        y_pred_fraud_weighted,
        target_names=["Non-Fraud", "Fraud"],
        zero_division=0
    )
)
```

```python
print("========== FRAUD THRESHOLD OPTIMIZATION ==========")

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

threshold_results_fraud = []

thresholds = np.arange(0.10, 0.91, 0.05)

for threshold in thresholds:

    y_pred_threshold = (
        y_prob_fraud_weighted >= threshold
    ).astype(int)

    threshold_results_fraud.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_test_fraud_fe,
            y_pred_threshold,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test_fraud_fe,
            y_pred_threshold,
            zero_division=0
        ),
        "F1": f1_score(
            y_test_fraud_fe,
            y_pred_threshold,
            zero_division=0
        )
    })

threshold_results_fraud = pd.DataFrame(
    threshold_results_fraud
)

display(
    threshold_results_fraud.round(4)
)
```

```python
print("========== BEST FRAUD THRESHOLD BY F1 ==========")

best_fraud_threshold = (
    threshold_results_fraud
    .loc[
        threshold_results_fraud["F1"].idxmax()
    ]
)

display(
    best_fraud_threshold.to_frame().T.round(4)
)
```

```python
print("========== FINAL FRAUD THRESHOLD EVALUATION ==========")

final_fraud_threshold_used = 0.55

y_pred_fraud_final = (
    y_prob_fraud_weighted >= final_fraud_threshold_used
).astype(int)

final_confusion_matrix_fraud = confusion_matrix(
    y_test_fraud_fe,
    y_pred_fraud_final
)

final_precision_fraud = precision_score(
    y_test_fraud_fe,
    y_pred_fraud_final,
    zero_division=0
)

final_recall_fraud = recall_score(
    y_test_fraud_fe,
    y_pred_fraud_final,
    zero_division=0
)

final_f1_fraud = f1_score(
    y_test_fraud_fe,
    y_pred_fraud_final,
    zero_division=0
)

print("Threshold:", final_fraud_threshold_used)

print("\nConfusion Matrix:")
print(final_confusion_matrix_fraud)

print("\nPrecision:",
      round(final_precision_fraud, 4))

print("Recall   :",
      round(final_recall_fraud, 4))

print("F1 Score :",
      round(final_f1_fraud, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_test_fraud_fe,
        y_pred_fraud_final,
        target_names=["Non-Fraud", "Fraud"],
        zero_division=0
    )
)
```

```python
print("========== FINAL FRAUD MODEL COMPARISON ==========")

fraud_model_comparison = pd.DataFrame({
    "Model": [
        "Baseline XGBoost",
        "Class-Weighted XGBoost",
        "Class-Weighted XGBoost (Threshold 0.55)"
    ],

    "Precision": [
        precision_fraud_baseline,
        precision_fraud_weighted,
        final_precision_fraud
    ],

    "Recall": [
        recall_fraud_baseline,
        recall_fraud_weighted,
        final_recall_fraud
    ],

    "F1": [
        f1_fraud_baseline,
        f1_fraud_weighted,
        final_f1_fraud
    ],

    "ROC_AUC": [
        roc_auc_fraud_baseline,
        roc_auc_fraud_weighted,
        roc_auc_fraud_weighted
    ],

    "PR_AUC": [
        pr_auc_fraud_baseline,
        pr_auc_fraud_weighted,
        pr_auc_fraud_weighted
    ]
})

display(
    fraud_model_comparison.round(4)
)
```

```python
print("========== FINAL FRAUD MODEL SELECTION ==========")

final_fraud_model_used = xgb_fraud_weighted
final_fraud_threshold_used = 0.55

print("Selected Model: Class-Weighted XGBoost")
print("Selected Threshold:", final_fraud_threshold_used)

print("\nPrecision:",
      round(final_precision_fraud, 4))

print("Recall   :",
      round(final_recall_fraud, 4))

print("F1 Score :",
      round(final_f1_fraud, 4))

print("ROC-AUC  :",
      round(roc_auc_fraud_weighted, 4))

print("PR-AUC   :",
      round(pr_auc_fraud_weighted, 4))
```
