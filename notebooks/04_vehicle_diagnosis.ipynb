```python
print("========== CREATING DATASET 4 FEATURE-ENGINEERED DATASET ==========")

# Renamed df4_fe to dataset4_fe to match your train-test split cell
dataset4_fe = df4.copy()

# Feature 1: Car brand
dataset4_fe["Car_Brand"] = (
    dataset4_fe["Car Name"]
    .str.split()
    .str[0]
)

# Feature 2: Car year
dataset4_fe["Car_Year"] = (
    dataset4_fe["Car Name"]
    .str.extract(r"(\d{4})")[0]
    .astype(int)
)

# Feature 3: Car age
dataset4_fe["Car_Age"] = (
    dataset4_fe["Car_Year"].max()
    - dataset4_fe["Car_Year"]
)

# Feature 4: Problem + Severity interaction
dataset4_fe["Problem_Severity"] = (
    dataset4_fe["Problem Classification"].astype(str)
    + "_"
    + dataset4_fe["Severity"].astype(str)
)

# Feature 5: Diagnosis + Severity interaction
dataset4_fe["Diagnosis_Severity"] = (
    dataset4_fe["Diagnosis"].astype(str)
    + "_"
    + dataset4_fe["Severity"].astype(str)
)

# Feature 6: ECU + Problem interaction
dataset4_fe["ECU_Problem"] = (
    dataset4_fe["ECU Data"].astype(str)
    + "_"
    + dataset4_fe["Problem Classification"].astype(str)
)

# Feature 7: Service history + Problem interaction
dataset4_fe["Service_Problem"] = (
    dataset4_fe["Service History"].astype(str)
    + "_"
    + dataset4_fe["Problem Classification"].astype(str)
)

print("Feature engineering completed successfully!")

print("\nOriginal features:", df4.shape[1])
print("New features:", dataset4_fe.shape[1] - df4.shape[1])
print("Total features:", dataset4_fe.shape[1])

print("\nNew features:")
for col in [
    "Car_Brand",
    "Car_Year",
    "Car_Age",
    "Problem_Severity",
    "Diagnosis_Severity",
    "ECU_Problem",
    "Service_Problem"
]:
    print("-", col)
```

```python
print("========== CORRECTING DATASET 4 TRAIN TEST SPLIT ==========")

from sklearn.model_selection import train_test_split

# Target
y_dataset4 = dataset4_fe["Problem Classification"]

# Features
X_dataset4 = dataset4_fe.drop(
    columns=["Problem Classification"]
)

X_train_dataset4, X_test_dataset4, y_train_dataset4, y_test_dataset4 = train_test_split(
    X_dataset4,
    y_dataset4,
    test_size=0.20,
    random_state=42,
    stratify=y_dataset4
)

print("Training data:")
print("X_train:", X_train_dataset4.shape)
print("y_train:", y_train_dataset4.shape)

print("\nTesting data:")
print("X_test :", X_test_dataset4.shape)
print("y_test :", y_test_dataset4.shape)

print("\nTraining target distribution:")
print(y_train_dataset4.value_counts())

print("\nTraining target percentages:")
print(
    y_train_dataset4.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nTesting target distribution:")
print(y_test_dataset4.value_counts())

print("\nTesting target percentages:")
print(
    y_test_dataset4.value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("========== CREATING DATASET 4 PREPROCESSING PIPELINE ==========")

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical features
dataset4_numeric_features = X_train_dataset4.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Categorical features
dataset4_categorical_features = X_train_dataset4.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical features:")
print(dataset4_numeric_features)

print("\nNumber of numerical features:",
      len(dataset4_numeric_features))

print("\nCategorical features:")
print(dataset4_categorical_features)

print("\nNumber of categorical features:",
      len(dataset4_categorical_features))


dataset4_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            dataset4_numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            dataset4_categorical_features
        )
    ]
)

print("\nDataset 4 preprocessing pipeline created successfully!")
```

```python
print("========== FITTING DATASET 4 PREPROCESSOR ==========")

X_train_dataset4_processed = dataset4_preprocessor.fit_transform(
    X_train_dataset4
)

print(
    "Processed training shape:",
    X_train_dataset4_processed.shape
)
```

```python
print("========== TRANSFORMING DATASET 4 TEST DATA ==========")

X_test_dataset4_processed = dataset4_preprocessor.transform(
    X_test_dataset4
)

print(
    "Processed testing shape:",
    X_test_dataset4_processed.shape
)
```

```python
print("========== DATASET 4 PREPROCESSING VALIDATION ==========")

print(
    "Original training features:",
    X_train_dataset4.shape[1]
)

print(
    "Processed training features:",
    X_train_dataset4_processed.shape[1]
)

print(
    "\nOriginal testing features:",
    X_test_dataset4.shape[1]
)

print(
    "Processed testing features:",
    X_test_dataset4_processed.shape[1]
)

print(
    "\nTraining samples:",
    X_train_dataset4_processed.shape[0]
)

print(
    "Testing samples:",
    X_test_dataset4_processed.shape[0]
)

print("\nTraining target classes:")
print(y_train_dataset4.unique())

print("\nTesting target classes:")
print(y_test_dataset4.unique())
```

```python
print("========== CREATING BASELINE DATASET 4 XGBOOST ==========")

from xgboost import XGBClassifier

dataset4_classes = sorted(y_train_dataset4.unique())

dataset4_class_to_int = {
    cls: i for i, cls in enumerate(dataset4_classes)
}

dataset4_int_to_class = {
    i: cls for cls, i in dataset4_class_to_int.items()
}

y_train_dataset4_encoded = y_train_dataset4.map(
    dataset4_class_to_int
)

y_test_dataset4_encoded = y_test_dataset4.map(
    dataset4_class_to_int
)

xgb_dataset4_baseline = XGBClassifier(
    objective="multi:softprob",
    num_class=len(dataset4_classes),
    n_estimators=200,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="mlogloss",
    n_jobs=-1
)

print("Classes:")
print(dataset4_classes)

print("\nClass mapping:")
print(dataset4_class_to_int)

print("\nBaseline Dataset 4 XGBoost created successfully!")
```

```python
print("========== TRAINING BASELINE DATASET 4 XGBOOST ==========")

import time

start_time = time.time()

xgb_dataset4_baseline.fit(
    X_train_dataset4_processed,
    y_train_dataset4_encoded
)

training_time_dataset4_baseline = time.time() - start_time

print("Training completed successfully!")
print(
    "Training time:",
    round(training_time_dataset4_baseline, 2),
    "seconds"
)
```

```python
print("========== DATASET 4 BASELINE XGBOOST PREDICTIONS ==========")

y_pred_dataset4_baseline = xgb_dataset4_baseline.predict(
    X_test_dataset4_processed
)

y_prob_dataset4_baseline = xgb_dataset4_baseline.predict_proba(
    X_test_dataset4_processed
)

print("Predictions generated successfully!")

print("\nFirst 20 encoded predictions:")
print(y_pred_dataset4_baseline[:20])

print("\nFirst 20 predicted classes:")
print([
    dataset4_int_to_class[int(x)]
    for x in y_pred_dataset4_baseline[:20]
])

print("\nFirst 5 probability rows:")
print(y_prob_dataset4_baseline[:5])
```

```python
print("========== CREATING LEAKAGE-FREE DATASET 4 FEATURES ==========")

dataset4_fe_clean = df4.copy()

# Feature 1: Car brand
dataset4_fe_clean["Car_Brand"] = (
    dataset4_fe_clean["Car Name"]
    .str.split()
    .str[0]
)

# Feature 2: Car year
dataset4_fe_clean["Car_Year"] = (
    dataset4_fe_clean["Car Name"]
    .str.extract(r"(\d{4})")[0]
    .astype(int)
)

# Feature 3: Car age
dataset4_fe_clean["Car_Age"] = (
    dataset4_fe_clean["Car_Year"].max()
    - dataset4_fe_clean["Car_Year"]
)

# Feature 4: Diagnosis + Severity interaction
# Does NOT directly contain Problem Classification
dataset4_fe_clean["Diagnosis_Severity"] = (
    dataset4_fe_clean["Diagnosis"].astype(str)
    + "_"
    + dataset4_fe_clean["Severity"].astype(str)
)

print("Feature engineering completed successfully!")

print("\nOriginal features:", df4.shape[1])
print(
    "New leakage-free features:",
    dataset4_fe_clean.shape[1] - df4.shape[1]
)
print(
    "Total features:",
    dataset4_fe_clean.shape[1]
)

print("\nFeatures added:")
for col in [
    "Car_Brand",
    "Car_Year",
    "Car_Age",
    "Diagnosis_Severity"
]:
    print("-", col)

print("\nRemoved leakage features:")
for col in [
    "Problem_Severity",
    "ECU_Problem",
    "Service_Problem"
]:
    print("-", col)
```

```python
print("========== DATASET 4 LEAKAGE-FREE TRAIN TEST SPLIT ==========")

from sklearn.model_selection import train_test_split

# Separate target
X_dataset4_clean = dataset4_fe_clean.drop(
    columns=["Problem Classification"]
)

y_dataset4_clean = dataset4_fe_clean[
    "Problem Classification"
]

# Stratified 80/20 split
X_train_dataset4, X_test_dataset4, y_train_dataset4, y_test_dataset4 = (
    train_test_split(
        X_dataset4_clean,
        y_dataset4_clean,
        test_size=0.20,
        random_state=42,
        stratify=y_dataset4_clean
    )
)

print("Training data:")
print("X_train:", X_train_dataset4.shape)
print("y_train:", y_train_dataset4.shape)

print("\nTesting data:")
print("X_test :", X_test_dataset4.shape)
print("y_test :", y_test_dataset4.shape)

print("\nTraining target distribution:")
print(y_train_dataset4.value_counts())

print("\nTraining target percentages:")
print(
    y_train_dataset4.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nTesting target distribution:")
print(y_test_dataset4.value_counts())

print("\nTesting target percentages:")
print(
    y_test_dataset4.value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
print("========== CREATING LEAKAGE-FREE DATASET 4 PREPROCESSING PIPELINE ==========")

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Numerical features
dataset4_numerical_features_clean = [
    "Car_Year",
    "Car_Age"
]

# Categorical features
dataset4_categorical_features_clean = [
    col for col in X_train_dataset4.columns
    if col not in dataset4_numerical_features_clean
]

print("Numerical features:")
print(dataset4_numerical_features_clean)

print(
    "\nNumber of numerical features:",
    len(dataset4_numerical_features_clean)
)

print("\nCategorical features:")
print(dataset4_categorical_features_clean)

print(
    "\nNumber of categorical features:",
    len(dataset4_categorical_features_clean)
)

# Numerical preprocessing
dataset4_numeric_pipeline_clean = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

# Categorical preprocessing
dataset4_categorical_pipeline_clean = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

# Combined preprocessing
dataset4_preprocessor_clean = ColumnTransformer(
    transformers=[
        (
            "num",
            dataset4_numeric_pipeline_clean,
            dataset4_numerical_features_clean
        ),
        (
            "cat",
            dataset4_categorical_pipeline_clean,
            dataset4_categorical_features_clean
        )
    ]
)

print("\nLeakage-free Dataset 4 preprocessing pipeline created successfully!")
```

```python
print("========== FITTING LEAKAGE-FREE DATASET 4 PREPROCESSOR ==========")

X_train_dataset4_processed_clean = (
    dataset4_preprocessor_clean.fit_transform(
        X_train_dataset4
    )
)

print(
    "Processed training shape:",
    X_train_dataset4_processed_clean.shape
)
```

```python
print("========== TRANSFORMING LEAKAGE-FREE DATASET 4 TEST DATA ==========")

X_test_dataset4_processed_clean = (
    dataset4_preprocessor_clean.transform(
        X_test_dataset4
    )
)

print(
    "Processed testing shape:",
    X_test_dataset4_processed_clean.shape
)
```
